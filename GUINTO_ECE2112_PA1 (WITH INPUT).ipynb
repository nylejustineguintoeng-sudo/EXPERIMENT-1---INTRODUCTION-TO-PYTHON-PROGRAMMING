### A. WORD ROTATION PROBLEM

```python
# Function to rotate a string by moving its first character to the end
def rotate_word(text: str):
    return text[1:] + text[0]


user_input = input("Enter a word: ")

if user_input:
    result = rotate_word(user_input)
    print(f"Rotated word: {result}")
else:
    print("Please enter a non-empty string.")
```

### B. USERNAME BUILDER PROBLEM

```python
# Function to generate a formatted username using basic string methods and concatenation
def make_username(first_name: str, last_name: str):
    n_first = first_name.lower().replace(" ", "")
    n_last = last_name.lower().replace(" ", "")
    return n_first + "." + n_last


# Get user inputs
first_input = input("Enter first name: ")
last_input = input("Enter last name: ")

if first_input and last_input:
    username = make_username(first_input, last_input)
    print(f"Generated username: {username}")
else:
    print("Please enter both first and last names.")
```

### C. BOOKEND SWAP PROBLEM

```python
# Function to swap the first and last elements using extended sequence unpacking
def swap_bookends(items: list):
    first, *middle, last = items
    return [last] + middle + [first]


# Get user input (separate elements with spaces)
user_input = input("Enter list elements separated by spaces: ")

# Convert the input string into a list of elements
items_list = user_input.split()

# Check if the list has at least two elements
if len(items_list) >= 2:
    result = swap_bookends(items_list)
    print(f"Swapped list: {result}")
else:
    print("Please enter at least two elements.")
```
